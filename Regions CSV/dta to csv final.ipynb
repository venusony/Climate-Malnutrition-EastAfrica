To convert Stata `.dta` files to `.csv` I followed an approach from Stack Overflow,  
using `pandas.read_stata()` and `.to_csv()` as suggested by LondonRob (2014).  
Source: https://stackoverflow.com/questions/2536047/convert-a-dta-file-to-csv-without-stata-software


LondonRob. “Convert a dta file to csv without Stata software.” *Stack Overflow*.  
https://stackoverflow.com/questions/2536047/convert-a-dta-file-to-csv-without-stata-software

```python
import pandas as pd
from pandas.io.stata import StataReader

# I am using one reference file to build my variable dictionary
ref_dta = "Kenya_KR_2022.dta"

with StataReader(ref_dta) as r:
    var_labels = r.variable_labels()

rename_map = {}

for var, lab in var_labels.items():
    # if there is no label, skip and leave it as code later
    if not isinstance(lab, str) or lab.strip() == "":
        continue
    
    new_name = lab.strip()
    
    # small cleanups so it matches how I would write it
    new_name = new_name.replace("standard deviations (according to WHO)", "SD per WHO")
    new_name = new_name.replace("standard deviation (according to WHO)", "SD per WHO")
    new_name = new_name.replace("standard deviations from the reference median", "SD from reference median")
    new_name = new_name.replace("standard deviation from the reference median", "SD from reference median")
    
    # remove extra spaces
    new_name = " ".join(new_name.split())
    
    rename_map[var] = new_name

# quick check of the first few entries
for k in list(rename_map.keys())[:20]:
    print(k, "->", rename_map[k])
```

```python
import pandas as pd

# all the csv files I made from the dta files
csv_files = [
    "Ethiopia_KR_1992.csv",
    "Ethiopia_KR_1997.csv",
    "Ethiopia_KR_2003.csv",
    "Ethiopia_KR_2008.csv",
    "Ethiopia_KR_2011.csv",
    "Kenya_KR_89.csv",
    "Kenya_KR_93.csv",
    "Kenya_KR_98.csv",
    "Kenya_KR_2003.csv",
    "Kenya_KR_2009.csv",
    "Kenya_KR_2014.csv",
    "Kenya_KR_2022.csv",
    "Tanzania_KR_92.csv",
    "Tanzania_KR_96.csv",
    "Tanzania_KR_99.csv",
    "Tanzania_KR_2004.csv",
    "Tanzania_KR_2010.csv",
    "Tanzania_KR_2015.csv",
    "Tanzania_KR_2022.csv",
    "Uganda_KR_88.csv",
    "Uganda_KR_95.csv",
    "Uganda_KR_2001.csv",
    "Uganda_KR_2006.csv",
    "Uganda_KR_2011.csv",
    "Uganda_KR_2016.csv",
]

for file in csv_files:
    print("working on:", file)
    df = pd.read_csv(file)

    # copy the sample weight into a simpler column
    if "v005" in df.columns and "weight_raw" not in df.columns:
        df["weight_raw"] = df["v005"]

    # make a scaled weight for later analysis
    if "weight_raw" in df.columns and "weight" not in df.columns:
        df["weight"] = df["weight_raw"] / 1000000.0

    clean_name = file.replace(".csv", "_clean.csv")
    df.to_csv(clean_name, index=False)
    print("saved:", clean_name)

print("done making clean csv files")
```

```python
# basic rename map part one
rename_basic = {
    # survey and ids
    "v000": "survey_code",
    "v001": "cluster",
    "v002": "household",
    "v003": "mother_line",
    "v004": "ultimate area unit",
    "v005": "weight_raw",
    "v006": "month of interview",
    "v007": "year_interview",
    "v008": "date of interview CMC",

    # mother background
    "v012": "mother_age_years",
    "v013": "mother age group",
    "v106": "mother_education_level",
    "v190": "wealth_quintile",

    # geography
    "v101": "region",
    "v102": "residence_type",

    # child info
    "bidx": "birth index",
    "bord": "birth order",
    "b4": "child_sex",
    "b5": "child_alive",

    # child age and measures
    "hw1": "child_age_months",
    "hw2": "child_height_cm",
    "hw3": "child_weight_kg",

    # z scores rough names for now
    "hw70": "z_height_for_age",
    "hw71": "z_weight_for_age",
    "hw72": "z_weight_for_height",
}

csv_clean_files = [
    "Ethiopia_KR_1992_clean.csv",
    "Ethiopia_KR_1997_clean.csv",
    "Ethiopia_KR_2003_clean.csv",
    "Ethiopia_KR_2008_clean.csv",
    "Ethiopia_KR_2011_clean.csv",
    "Kenya_KR_89_clean.csv",
    "Kenya_KR_93_clean.csv",
    "Kenya_KR_98_clean.csv",
    "Kenya_KR_2003_clean.csv",
    "Kenya_KR_2009_clean.csv",
    "Kenya_KR_2014_clean.csv",
    "Kenya_KR_2022_clean.csv",
    "Tanzania_KR_92_clean.csv",
    "Tanzania_KR_96_clean.csv",
    "Tanzania_KR_99_clean.csv",
    "Tanzania_KR_2004_clean.csv",
    "Tanzania_KR_2010_clean.csv",
    "Tanzania_KR_2015_clean.csv",
    "Tanzania_KR_2022_clean.csv",
    "Uganda_KR_88_clean.csv",
    "Uganda_KR_95_clean.csv",
    "Uganda_KR_2001_clean.csv",
    "Uganda_KR_2006_clean.csv",
    "Uganda_KR_2011_clean.csv",
    "Uganda_KR_2016_clean.csv",
]

for fname in csv_clean_files:
    print("working on:", fname)
    df = pd.read_csv(fname)

    cols = df.columns.tolist()

    local_map = {}
    for old, new in rename_basic.items():
        if old in cols:
            local_map[old] = new

    print("  renaming", len(local_map), "columns in this step")

    df = df.rename(columns=local_map)
    df.to_csv(fname, index=False)

print("done basic relabel part one")
```

```python
from pathlib import Path

# this is where my unzipped DHS folders live
base = Path(r"C:\Users\venus\OneDrive\Documents\GitHub\Climate-Malnutrition-EastAfrica\data_raw\unzipped")

# I am mapping the old DHS folder codes to readable names
folder_rename = {
    "ETKR41DT": "Ethiopia_KR_2005",
    "ETKR51DT": "Ethiopia_KR_51",
    "ETKR61DT": "Ethiopia_KR_2011",
    "ETKR71DT": "Ethiopia_KR_2016",
    "ETKR81DT": "Ethiopia_KR_2021",

    "KEKR01DT": "Kenya_KR_1988",
    "KEKR31DT": "Kenya_KR_1998",
    "KEKR3ADT": "Kenya_KR_1999",
    "KEKR42DT": "Kenya_KR_2008",
    "KEKR52DT": "Kenya_KR_2010",
    "KEKR72DT": "Kenya_KR_72",    # keep this label for now
    "KEKR8CDT": "Kenya_KR_2022",

    "TZKR21DT": "Tanzania_KR_1996",
    "TZKR3ADT": "Tanzania_KR_1999",
    "TZKR41DT": "Tanzania_KR_2005",
    "TZKR4IDT": "Tanzania_2007_KR",
    "TZKR63DT": "Tanzania_KR_2013",
    "TZKR7BDT": "Tanzania_KR_2018",
    "TZKR82DT": "Tanzania_KR_2023",

    "UGKR01DT": "Uganda_KR_1988",
    "UGKR33DT": "Uganda_KR_2000",
    "UGKR41DT": "Uganda_KR_2005",
    "UGKR52DT": "Uganda_KR_2010",
    "UGKR61DT": "Uganda_KR_2011",
    "UGKR7BDT": "Uganda_KR_2018",
}

for old_name, new_name in folder_rename.items():
    old_path = base / old_name
    new_path = base / new_name

    if not old_path.exists():
        print(f"skipping {old_name} because I do not see that folder")
        continue

    if new_path.exists():
        print(f"skipping {old_name} because {new_name} already exists")
        continue

    old_path.rename(new_path)
    print(f"renamed folder {old_name} → {new_name}")
```

```python
from pathlib import Path

# this is where all my renamed country folders live
# example folders in here: Ethiopia_KR_2005, Kenya_KR_2022, Uganda_KR_2018, etc
unzipped_root = Path(r"C:\Users\venus\OneDrive\Documents\GitHub\Climate-Malnutrition-EastAfrica\data_raw\unzipped")

# file types I care about for DHS support files
target_exts = {".dta", ".dct", ".do", ".frq", ".frw", ".map"}

for folder in sorted(unzipped_root.iterdir()):
    if not folder.is_dir():
        continue

    nice_base = folder.name  # this is already Country_KR_Year which is what I want
    print(f"\ninside folder: {nice_base}")

    for f in folder.iterdir():
        if not f.is_file():
            continue

        ext = f.suffix.lower()

        # only touch the DHS files I actually care about
        if ext in target_exts:
            new_name = f"{nice_base}{ext}"
            new_path = folder / new_name

            # only rename if the name is different
            if f.name != new_name:
                print(f"  renaming {f.name} -> {new_name}")
                f.rename(new_path)
            else:
                print(f"  keeping {f.name} (already clean)")

print("\nfinished renaming inner DHS files.")
```
